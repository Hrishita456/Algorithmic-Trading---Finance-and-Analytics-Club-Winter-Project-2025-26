```python
"""Question 1
The first question involves arithmetic with a small application of loops.
You are given an array of n integers, that is, a1, a2, . . . , an,
and you are required to output the maximum and minimum among them."""

def max_list(l):
  max = l[0]
  for i in l:
    if i>max:
      max = i
  return max

def min_list(l):
  min = l[0]
  for i in l:
    if i<min:
      min = i
  return min


```

```python
"""Question 2
This question is for familiarity with strings.
You are needed to input a string and check if
it is a palindrome."""

def check_palindrome(s):
  left = 0
  right = len(s) - 1
  a = True
  while left < right:
    if s[left]==s[right]:
      left += 1
      right-= 1
    else:
      a = False
      break

  if a == True:
    print("It is a palindrome")
  else:
    print("It is not a palindrome")

check_palindrome("anna")
check_palindrome("ann")

```

```python
"""Question 3
You will need a basic understanding of sorting for this question.
Note that there is a direct Python function to sort a sequence of integers.
You are given an array of n integers, that is, a1, a2, . . . , an, and a number x.
You need to find a pair of elements whose absolute difference is x.
You are assured that such a pair exists in the array."""

def find_pair_with_diff(arr, x):
  arr.sort()
  left = 0
  right = 1
  while right<len(arr):
    diff = arr[right]-arr[left]
    if diff==x:
      return (arr[right],arr[left])
    elif diff>x:
      left +=1
    else:
      right +=1
    if right == left:
      right +=1

print(find_pair_with_diff([1,2,3,4,8],3))

```
